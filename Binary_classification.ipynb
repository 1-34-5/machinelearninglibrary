```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
```

```python
#load the training and test data 
data_train = pd.read_csv('/home/bigbang/machinelearning_algorithms/binary_classification_train.csv')
X_train = data_train.iloc[:,1:-1].values
y = data_train.iloc[: , -1].values 

data_test = pd.read_csv('/home/bigbang/machinelearning_algorithms/binary_classification_test.csv')
X_test = data_test.iloc[:, 1:].values 

# Normalize training data
X_train_mean = X_train.mean(axis=0)
X_train_std = X_train.std(axis=0)

X = (X_train - X_train_mean) / X_train_std  # Standardize training features

# Normalize test data using training data's mean and std
X_test_normalized = (X_test - X_train_mean) / X_train_std

print("Test data after normalization:")
print(X_test_normalized[:5])  # Print first 5 rows of normalized test data



# Display the first few rows of the training data to confirm it's loaded correctly
print(X[:5])
print(y[:5])
```

```python
def LogReg_with_gradient_descent(X, y , alpha , epoch):
    m = X.shape[0]
    ones = np.ones((m,1))
    X = np.concatenate((ones,X), axis=1)
    n = X.shape[1]
    w = np.zeros(n)
    costs = []

    def sigmoid(z):
        return 1/(1 + np.exp(-z))

    def compute_cost(X, y, w):
        m = len(y)
        h = sigmoid(np.dot(X,w))
        cost = -(1 /m) * np.sum(y * np.log(h) + (1 - y) * np.log(1-h))
        return cost 

    for i in range(epoch):
        h = sigmoid(np.dot(X,w))
        gradient = (1/m) * np.dot(X.T, (h-y))
        w = w - alpha * gradient

        cost = compute_cost(X,y,w)
        costs.append(cost)
        if i % 100 == 0:
            print(f'Epoch {i}, Cost: {cost}')

    return w,costs 
```

```python
alpha = 0.01  # Learning rate
epoch = 1000  # Number of iterations
weights,costs = LogReg_with_gradient_descent(X,y,alpha,epoch)
print("Learned weights:" , weights)

def predict(X,w):
    m = X.shape[0]
    ones = np.ones((m,1))
    X = np.concatenate((ones,X), axis = 1)
    probabilities = 1/(1+np.exp(-np.dot(X,w)))
    predictions = (probabilities >= 0.5).astype(int)
    return predictions 

y_pred = predict(X,weights)
print(y_pred[:5])
```

```python
#checking the errors in prediction of training data 
count = 0
m = X.shape[0]
for i in range(m):
    if(y[i] != y_pred[i]):
        print(f'Error at the index {i}')
        count += 1
    else:
        continue 

print(f'Number of errors {count}')
```

```python
def calculate_f1_score(y, y_pred):
    TP = np.sum((y == 1) & (y_pred == 1))
    FP = np.sum((y == 0) & (y_pred == 1))
    FN = np.sum((y == 1) & (y_pred == 0))

    precision = TP/ (TP + FP) if (TP + FP) > 0 else 0 
    recall = TP/ (TP + FN) if (TP + FN) > 0 else 0

    f1_score = 2 * (precision * recall) / (precision + recall) 
    return precision, recall, f1_score 
    
precision,recall,f1_score = calculate_f1_score(y, y_pred)
print(f'f1_score {f1_score}')  
```
